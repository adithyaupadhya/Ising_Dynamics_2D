# Droplet evolution

Initialize a circular plus droplet using the public set_state method; its geometry is an initial condition, not an equilibrium sample.

Run cells in order. Install the package with the notebook extra first (see README).

```python
import numpy as np
import matplotlib.pyplot as plt
from ising import IsingModel, Simulator
from ising.experiments.trajectory import run
```

```python
N=48
sim=Simulator(IsingModel(N,.6,4.),seed=42)
y,x=np.ogrid[:N,:N]
state=sim.state.copy()
state[1:-1,1:-1][(x-(N-1)/2)**2+(y-(N-1)/2)**2<=(N/4)**2]=1
sim.set_state(state)
fig,axes=plt.subplots(1,4,figsize=(12,3)); rows=[]
for j,ax in enumerate(axes):
    if j: sim.advance(25)
    rows.append(sim.observe(contours=True,interface_heights=True))
    ax.imshow(sim.state,cmap='coolwarm',vmin=-1,vmax=1); ax.set_title(f't={sim.t:g}'); ax.axis('off')
plt.show()
plt.plot([r['sweep'] for r in rows],[r['largest_plus_fraction'] for r in rows],'o-'); plt.xlabel('Sweeps'); plt.ylabel('Largest plus fraction'); plt.show()
print(rows[-1])
```
