# Exact equilibrium validation

For N ≤ 4 enumerate all configurations. Short Monte Carlo runs and batch standard errors are diagnostics, not guarantees.

Run cells in order. Install the package with the notebook extra first (see README).

```python
import numpy as np
import matplotlib.pyplot as plt
from ising import IsingModel, Simulator
from ising.experiments.trajectory import run
```

```python
from ising.exact import exact_small
from ising.experiments.equilibrium import equilibrium_check
model=IsingModel(3,.4,2.)
w,bonds,msum=exact_small(model.N,model.beta,model.lam)
assert np.isclose(w.sum(),1)
values=np.unique(msum)
plt.bar(values/model.N**2,[w[msum==v].sum() for v in values]); plt.xlabel('Magnetization'); plt.ylabel('Exact probability'); plt.show()
exact,rows=equilibrium_check(model,steps=1000,burn=100,seed=42)
print(exact)
for r in rows: print(r)
```
