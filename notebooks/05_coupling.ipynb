# Monotone coupling

Extremal heat-bath chains share random updates. Coalescence is a coupling-time observation, not an exact mixing time.

Run cells in order. Install the package with the notebook extra first (see README).

```python
import numpy as np
import matplotlib.pyplot as plt
from ising import IsingModel, Simulator
from ising.experiments.trajectory import run
```

```python
from ising import MonotoneCoupling
pair=MonotoneCoupling(IsingModel(12,.45,3.),42)
rows=[pair.observe()]
for _ in range(200):
    if pair.coalesced: break
    pair.advance(); rows.append(pair.observe())
assert np.all(pair.lower<=pair.upper)
plt.plot([r['sweep'] for r in rows],[r['disagreement_fraction'] for r in rows]); plt.xlabel('Sweeps'); plt.ylabel('Disagreement fraction'); plt.show()
print('Coalesced:',pair.coalesced,'Observed time:',pair.t)
```
