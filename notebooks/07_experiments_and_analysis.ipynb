# Replicated first-passage experiments

Trials not crossing the threshold before the limit are right-censored. Use Kaplan–Meier curves. Small samples are illustrative only.

Run cells in order. Install the package with the notebook extra first (see README).

```python
import numpy as np
import matplotlib.pyplot as plt
from ising import IsingModel, Simulator
from ising.experiments.trajectory import run
```

```python
from ising.experiments.first_passage import first_passage
from ising.analysis.survival import kaplan_meier,median_survival
result=first_passage([8,12],beta=.5,lam=3.,limit=100,replicates=12,seed=42,workers=1)
for N in [8,12]:
    rows=[r for r in result.rows if r['N']==N]
    curve=kaplan_meier([r['sweeps'] for r in rows],[r['event'] for r in rows])
    plt.step([0]+[r['time'] for r in curve],[1]+[r['survival'] for r in curve],where='post',label=f'N={N}')
    print(N,'events:',sum(r['event'] for r in rows),'median:',median_survival(curve))
plt.xlabel('Sweeps'); plt.ylabel('Survival probability'); plt.legend(); plt.show()
```
