# Competing field and prewetting

At h=λ/N, bulk gain and macroscopic interface cost are both order N. This finite-time scan is not an equilibrium estimate or proof of metastability.

Run cells in order. Install the package with the notebook extra first (see README).

```python
import numpy as np
import matplotlib.pyplot as plt
from ising import IsingModel, Simulator
from ising.experiments.trajectory import run
```

```python
from ising.experiments.field_scan import field_scan
rows=field_scan(16,.55,[0.,2.,4.,6.],80,seed=42)
fig,axes=plt.subplots(2,2,figsize=(10,6),sharey=True)
for ax,k in zip(axes.flat,['heatbath','metropolis','sw-ghost','sw-size']):
    for init in ['minus','plus']:
        subset=[r for r in rows if r['dynamics']==k and r['init']==init]
        ax.plot([r['lam'] for r in subset],[r['mean_m'] for r in subset],'o-',label=init)
    ax.set(title=k,xlabel='lambda',ylabel='Late-time mean magnetization'); ax.legend()
fig.tight_layout(); plt.show()
```
